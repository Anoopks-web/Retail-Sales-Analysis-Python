KEY INSIGHTS

```python
# 1. Best-performing product categories
category_revenue = df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)
print("1. Best-performing Product Categories:")
print(category_revenue)
print()
```

```python
# 2. Highest-spending customers
customer_spending = df.groupby("Customer ID")["Total Amount"].sum().sort_values(ascending=False)
print("2. Highest-spending Customers:")
print(customer_spending.head(10))
print()
```

```python
# 3. Customer purchasing behavior
customer_behavior = df.groupby("Customer ID").agg(
    Total_Spending=("Total Amount", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Transactions=("Transaction ID", "count")
).sort_values("Total_Spending", ascending=False)

print("3. Customer Purchasing Behavior:")
print(customer_behavior.head(10))
print()
```

```python
# 4. Gender-wise purchasing patterns
gender_sales = df.groupby("Gender").agg(
    Total_Spending=("Total Amount", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Transactions=("Transaction ID", "count")
)

print("4. Gender-wise Purchasing Patterns:")
print(gender_sales)
print()

```

```python
# 5. Age-group spending patterns
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=["Under 18", "18-30", "31-45", "46-60", "60+"]
)
age_group_spending = df.groupby("Age Group", observed=True)["Total Amount"].sum()
print("5. Age-group Spending Patterns:")
print(age_group_spending)
print()

```

```python
# 6. Monthly and daily revenue trends
df["Date"] = pd.to_datetime(df["Date"])
monthly_revenue = df.groupby(df["Date"].dt.to_period("M"))["Total Amount"].sum()
daily_revenue = df.groupby("Date")["Total Amount"].sum()
print("6. Monthly Revenue:")
print(monthly_revenue)
print()
print("Daily Revenue:")
print(daily_revenue)
print()

```

```python
# 7. Relationship between quantity and spending
quantity_spending_corr = df["Quantity"].corr(df["Total Amount"])
print("7. Relationship between Quantity and Spending:")
print("Correlation:", round(quantity_spending_corr, 2))
print()
```

```python
# 8. Categories contributing most to overall revenue
category_contribution = (
    df.groupby("Product Category")["Total Amount"].sum()
    / df["Total Amount"].sum() * 100
).sort_values(ascending=False)

print("8. Category Contribution to Overall Revenue:")
print(category_contribution.round(2))
```
